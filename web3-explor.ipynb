### Normalize Event

```python
def normalize_supply_event(event):

    return {
        "block_number": event["blockNumber"],
        "tx_hash": event["transactionHash"].hex(),
        "market_id": event["args"]["id"],
        "supplier": event["args"]["onBehalf"],
        "assets": event["args"]["assets"],
        "shares": event["args"]["shares"]
    }
```

```python
rows = [
    normalize_supply_event(e)
    for e in events
]

print(rows[0])
```

```python
import pandas as pd

df = pd.DataFrame(rows)

df.head()
```
